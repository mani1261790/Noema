# 基本的なPythonの操作

Python の基礎は、値に名前を付け、入れ物へまとめ、条件で分け、繰り返し処理し、関数として再利用する流れです。文法を一覧で覚えるより、同じ小さなデータを少しずつ加工しながら、出力がどう変わるかを追います。

成績データを題材に、変数、リスト、辞書、集合、スライス、分岐、ループ、関数、例外処理を一続きの作業として扱います。

## まずは値に名前を付ける

プログラムの最初の一歩は、値をコンピュータへ預けることです。名前、人数、完了率のように、あとで使いたい値へ分かりやすい名前を付けます。


## 1 つの値をきちんと持つ

名前、受講人数、完了率のように性質の違う値を並べて持つと、Python が文字列、整数、小数をどう扱っているかが見えます。`type(...)` の表示は暗記するより、出力と値を対応づけて読みます。


`f"..."` は、値を文字列の中へ差し込む書き方です。`:.0%` を付けると、`0.45` を `45%` のような百分率表示へ整えます。値そのものを変えるのではなく、表示の形だけを変えています。

## 値が増えたら、入れ物を選ぶ

1 人分の情報なら変数だけでも足ります。点数が複数になると、目的に合った入れ物が必要になります。

- 順番を保ったまま並べたいときはリスト
- 項目名で取り出したいときは辞書
- 重複を消して「あるかどうか」だけ見たいときは集合

同じ題材でも入れ物を変えると、書きやすい処理が変わります。


同じ点数データを何度も眺め直します。一部を取り出し、条件でラベルを付け、関数で集計し、変換に失敗する値を処理します。ばらばらの機能ではなく、1 つのデータを扱いやすくする流れとして読みます。


## データを少しずつ加工する

インデックスとスライスは「どの位置の値を見るか」を決める道具です。分岐とループは「値ごとに別の処理をする」ための道具です。関数は、それらを名前付きのまとまりにして再利用しやすくするために使います。


```python
import math
from collections import Counter
```

`type` は、いま手元にある値が文字列なのか、整数なのか、小数なのかを確かめるための道具です。

Python は変数宣言の段階で型を書かなくても動くので、分からなくなったら実際に `type(...)` を出して確認する、という姿勢がかなり役に立ちます。

```python
name = "Noema"
students = 32
completion_ratio = 0.45

print(type(name), type(students), type(completion_ratio))
print(f"{name}: {students}人, 完了率 {completion_ratio:.0%}")
```

## 値を更新して表示も変える

変数は一度作って終わりではありません。値を更新すると、同じ表示式でも結果が変わります。計算結果を別の変数へ入れると、元の値と新しい値の違いも追いやすくなります。

```python
additional_students = 5
students_after_campaign = students + additional_students
new_completion_ratio = 0.62

print(f"追加後: {students_after_campaign}人")
print(f"完了率の変化: {completion_ratio:.0%} -> {new_completion_ratio:.0%}")
```

リスト、辞書、集合は、同じデータを別の角度から持つための入れ物です。`scores` は点数を順番つきで並べたもの、`profile` は名前付きの情報、`unique_scores` は重複を除いた点数の集合です。集合は表示順が固定ではないので、順番ではなく重複が消えたことを見ます。

```python
scores = [72, 88, 65, 91, 88]
profile = {"name": "Aiko", "grade": 1, "club": "AI"}
unique_scores = set(scores)

print("scores:", scores)
print("profile['club']:", profile["club"])
print("unique_scores:", unique_scores)
```

## 入れ物ごとの得意な操作を見る

リストは順番、辞書は名前付きの取り出し、集合は重複の確認に向いています。どの入れ物を使うかは、あとで何をしたいかで決まります。

```python
print("first score:", scores[0])
print("profile keys:", list(profile.keys()))
print("has 88?:", 88 in unique_scores)
print("duplicate count:", len(scores) - len(unique_scores))
```

インデックスは 1 個の値を見る操作、スライスは連続した範囲を見る操作です。

`start:end` の `end` が含まれないのは少し癖がありますが、そのおかげで『先頭から何個』という感覚で範囲を切り出しやすくなっています。`scores[1:4]` なら、1 番目から始めて 4 の手前で止まる、と読めば混乱が減ります。

```python
print("先頭:", scores[0])
print("末尾:", scores[-1])
print("2番目から4番目手前:", scores[1:4])
print("2個おき:", scores[::2])
```

## enumerate で位置と値を同時に見る

リストを処理するとき、値だけでなく何番目かも必要になることがあります。`enumerate` を使うと、位置と値を同時に取り出せます。

```python
for index, score in enumerate(scores):
    print(index, score)
```

`scores` の点数を A/B/C に振り分ける小さな判定表を作ります。ループは点数を 1 個ずつ取り出し、分岐はその点数に応じて付けるラベルを決めます。`zip(scores, labels)` で、元の点数と作ったラベルを並べ直して確認します。

```python
labels = []
for score in scores:
    if score >= 90:
        labels.append("A")
    elif score >= 75:
        labels.append("B")
    else:
        labels.append("C")

print(list(zip(scores, labels)))
```

## 判定ルールを関数へ分ける

同じ判定を何度も使うなら、分岐を関数に閉じ込めます。境界値を変えたいときも、関数の中だけを見ればよくなります。

```python
def grade_label(score):
    if score >= 90:
        return "A"
    if score >= 75:
        return "B"
    return "C"

labels_by_function = [grade_label(score) for score in scores]
print(labels_by_function)
print(labels_by_function == labels)
```

リスト内包表記は、`for` で新しいリストを作る処理を短く書いた形です。普通の `for` と同じ出力になることを確認してから、1 行の書き方へ移ると安全です。

```python
# 通常の for
squared_for = []
for x in range(1, 6):
    squared_for.append(x * x)

# 内包表記
squared_comp = [x * x for x in range(1, 6)]

print(squared_for)
print(squared_comp)
```

## 条件付きの内包表記で必要な値だけ残す

内包表記には条件も付けられます。合格点だけを取り出す、点数を補正する、といった処理を短く書けます。

```python
passed_scores = [score for score in scores if score >= 75]
curved_scores = [min(score + 5, 100) for score in scores]

print("passed:", passed_scores)
print("curved:", curved_scores)
```

関数は、何度も使う処理に名前を付ける仕組みです。`summarize_scores` は、点数の個数、平均、最高点、最低点をひとまとめに返します。分析コードでは、意味のあるまとまりを関数にしておくと、あとから読み返しやすくなります。

```python
def summarize_scores(values):
    count = len(values)
    avg = sum(values) / count
    best = max(values)
    worst = min(values)
    return {"count": count, "avg": avg, "best": best, "worst": worst}

summary = summarize_scores(scores)
print(summary)
```

## 関数は別のデータにも使える

関数にしておくと、同じ集計を別の点数リストにも使えます。処理がデータに依存しすぎていないかも確認できます。

```python
second_scores = [81, 77, 93, 69, 84]
print(summarize_scores(second_scores))
```

現実のデータは、いつもきれいに数値へ変換できるとは限りません。

`try` は『ここは失敗するかもしれない』と先に宣言する書き方で、`except ValueError` は『数値へ直せないという種類の失敗だけ受け取る』という意味です。エラーを黙って無視するのではなく、どこで失敗したかを出力しておくと、後で原因を追いやすくなります。

```python
raw_inputs = ["12", "7", "oops", "20"]
parsed = []

for item in raw_inputs:
    try:
        parsed.append(int(item))
    except ValueError:
        print(f"変換失敗: {item}")

print("parsed:", parsed)
```

## 失敗した値も記録する

変換できた値だけを残すと、何を捨てたのか分からなくなります。失敗した値を別のリストへ入れておくと、後で入力データを直す手掛かりになります。

```python
parsed = []
failed = []

for item in raw_inputs:
    try:
        parsed.append(int(item))
    except ValueError:
        failed.append(item)

print("parsed:", parsed)
print("failed:", failed)
```

最後は、ここまで作ってきたものを小さな集計としてまとめます。`Counter(labels)` は各ラベルが何回出たかを数えます。平均点は全体の中心、標準偏差は点数の散らばりを表します。平均だけでは見えないばらつきを、標準偏差で補います。

```python
label_counter = Counter(labels)
avg_score = sum(scores) / len(scores)
std_score = math.sqrt(sum((x - avg_score) ** 2 for x in scores) / len(scores))

print("label count:", label_counter)
print(f"average={avg_score:.2f}, std={std_score:.2f}")
```

## 集計を表に近い形へ整える

点数、ラベル、平均との差を同じ行に並べると、どの点が高いか低いかを読みやすくなります。Pandas を使う前でも、辞書のリストで小さな表を作れます。

```python
rows = []
for score, label in zip(scores, labels):
    rows.append({
        "score": score,
        "label": label,
        "diff_from_avg": round(score - avg_score, 2),
    })

for row in rows:
    print(row)
```

Python 基礎で大切なのは、値の置き方と処理の流れを自分で追えることです。点数の境界を変える、別の項目を足す、変換に失敗する文字列を増やすと、出力がどう変わるかを確認できます。
